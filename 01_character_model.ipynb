```python
chars = ['<S>', '<E>'] + sorted(list(set("".join(texts))))
chars
```

```python
import torch
```

```python
stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for ch, i in stoi.items()}

vocab_size = len(stoi)

N = torch.zeros((vocab_size, vocab_size), dtype=torch.int32)
vocab_size
```

```python
for text in texts:
    chs = ['<S>'] + list(text) + ['<E>']

    for ch1, ch2 in zip(chs, chs[1:]):
        i1 = stoi[ch1]
        i2 = stoi[ch2]
        N[i1][i2] += 1
```

```python
log_likelihood = 0
n =0
for text in texts:
    chs = ['<S>'] + list(text) + ['<E>']

    for ch1, ch2 in zip(chs, chs[1:]):
        i1 = stoi[ch1]
        i2 = stoi[ch2]
        prob = P[i1,i2]
        logprob = torch.log(prob)
        log_likelihood += logprob
        n+=1
print(f'log_likelihood: {log_likelihood}')
```

```python
nll = -log_likelihood
nll/n
```

```python
xs =[]
ys =[]
for text in texts:
    chs = ['<S>'] + list(text) + ['<E>']

    for ch1, ch2 in zip(chs, chs[1:]):
        i1 = stoi[ch1]
        i2 = stoi[ch2]
        N[i1][i2] += 1
        xs.append(i1)
        ys.append(i2)

xs = torch.tensor(xs)
ys = torch.tensor(ys)
```

```python

n = xs.nelement()
n
```

```python
ys.shape
```

```python
import torch.nn.functional as F
```

```python
g = torch.Generator().manual_seed(2147483647 + 2)
W = torch.randn((vocab_size,vocab_size), generator=g, requires_grad=True)
```

```python
for k in range(100):
    xenc = F.one_hot(xs, num_classes=vocab_size).float()
    logits = xenc@W
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdims=True)
    #loss
    loss = -probs[torch.arange(n), ys].log().mean() + 0.01* (W**2).mean()
    print(loss.item())
    #backward pass 
    W.grad = None
    loss.backward()
    #update
    W.data += -0.1*W.grad
```
